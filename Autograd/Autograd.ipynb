## finding dervative manually

```python
import torch 

#inputs
x=torch.tensor(6.7) #input feature
y=torch.tensor(0.0) #true label (binary)


w=torch.tensor(1.0) #weight
b=torch.tensor(0.0) #bias

```

```python
# Binary Cross Entropy Loss for Scalar
def binary_cross_entropy_loss(prediction,target):
    epsilon=1e-8 # to prevent log(0)
    prediction=torch.clamp(prediction,epsilon,1.0-epsilon)
    return -(target*torch.log(prediction)+(1-target)*torch.log(1-prediction))
```

```python
#forward pass
z=w*x+b #weghed sum (linear part)
y_pred=torch.sigmoid(z)  #predicted probability

#compute binary cross enbtroy lo9sss
loss=binary_cross_entropy_loss(y_pred,y)
```

```python
loss
```

```python
#derivative
# 1. dL/d(y_pred) : loss with repect to the prediction (y_pred)
dloss_dy_pred=(y_pred-y)/(y_pred*(1-y_pred))

#2.  dy_pred/dz :Prediction (y_pred) with repectr to z (sigmoid derivative)
dy_pred_dz=y_pred *(1-y_pred)

#3 dz/dw and dz/db :z ith respect to w and b
dz_dw=x #dz/dw=x
dz_db=1 #dz/db=1(bia contrivute directly to z)


dL_dw=dloss_dy_pred * dy_pred_dz * dz_dw
dL_db=dloss_dy_pred * dy_pred_dz * dz_db
```

```python
print(f"Manual Gradient of loss w.r.t weight (dw): {dL_dw}")
print(f"Manual Gradient of loss w.r.t bias (db): {dL_db}")
```

## finding derivative with AutoGrad

```python
x=torch.tensor(6.7)
y=torch.tensor(0.0)
```

```python
w=torch.tensor(1.0,requires_grad=True) #weight
b=torch.tensor(0.0,requires_grad=True) #bias
```

```python
w
```

```python
b
```

```python
z=w*x+b #weghed sum (linear part) 
z
```

```python
y_pred=torch.sigmoid(z)
y_pred
```

```python
loss=binary_cross_entropy_loss(y_pred,y)
```

```python
loss
```

```python
loss.backward()
```

```python
w.grad
```
