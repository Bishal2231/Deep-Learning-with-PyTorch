## Mathematical Operations


1. scalar Operation

```python
x=torch.rand(2,3)
x
```

```python
#addition
x+2

#subtraction
x-2

#multiplication
x*3

#division
x/3

#int division
(x*100)//3

#mod
((x*100)//3)%2

#power
x**2

```

2. Element Wise Operation

```python
a=torch.rand(2,3)
b=torch.rand(2,3)

print("a->\n",a)
print("b->\n",b)
```

```python
#add
a+b

#sub
a-b
#mul
a*b

#div
a/b

#power
a **b

#mod
a%b
```

```python
c=torch.tensor([1,-2,3,-4])
```

```python
#  abs -aBSOLUTE VALYUE
torch.abs(c)
```

```python
#negative
torch.neg(c)
```

```python
d=torch.tensor([1.9,2.3,3.7,4.4])
```

```python
#round
torch.round(d)
```

```python
#cell
torch.ceil(d)
```

```python
torch.floor(d)
```

```python
#clamp
torch.clamp(d,min=2,max=3)
```
